That’s where __slots__ comes in. Unlike __dict__, __slots__ 𝐩𝐫𝐞𝐯𝐞𝐧𝐭𝐬 𝐭𝐡𝐞 𝐜𝐫𝐞𝐚𝐭𝐢𝐨𝐧 𝐨𝐟 𝐚𝐧 𝐢𝐧𝐬𝐭𝐚𝐧𝐜𝐞 𝐝𝐢𝐜𝐭𝐢𝐨𝐧𝐚𝐫𝐲, meaning you won’t be able to add attributes to objects at any time. This, in contrast, is a way of implementing immutable objects in Python.

What advantages does this bring to your code?
🚀 Faster attribute access: not at instantiation time, but when r𝐫𝐞𝐭𝐫𝐢𝐞𝐯𝐢𝐧𝐠 𝐚𝐧𝐝 𝐬𝐞𝐭𝐭𝐢𝐧𝐠  values.
🚀 Saving tons of space in memory: an object with empty __slots__ takes up only 32 bytes in total, 40 bytes with one item. Without __slots__, an object with __dict__ requires 56 bytes. However, since __dict__ is only created when referenced, it can spike the class instance size to a total of 296 𝐛𝐲𝐭𝐞𝐬.

🥁 To provide evidence, I created 10,000 objects using __dict__ and another 10,000 using __slots__. The results were as expected—I saved almost 400 𝐊𝐁 of memory by using __slots__.

```python
import sys

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

class SlottedPoint:
    __slots__ = ['x', 'y']
    
    def __init__(self, x, y):
        self.x = x
        self.y = y

# Create instances
p1 = Point(1, 2)
p2 = SlottedPoint(1, 2)

# Compare memory usage
print("Point object size:", sys.getsizeof(p1) + sys.getsizeof(p1.__dict__)) 
print("SlottedPoint object size:", sys.getsizeof(p2))  
```

Since SlottedPoint avoids storing attributes in __dict__, it uses less memory, making it more efficient for large-scale object creation.
