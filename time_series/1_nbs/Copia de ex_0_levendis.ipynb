```python
# 1
def get_lag(lista,lag):
    return  lista[lag:] + [np.nan for i in range(lag)]

# Given time series
X = [2, 4, 6, 8, 10]

# Calculate first and second lags
first_lag = get_lag(X,1)
second_lag = get_lag(X,2)

# Display the results
print("Original Series:", X)
print("First Lag:", first_lag)
print("Second Lag:", second_lag)
```

```python
def get_diff(X, n=1):
    diffs = np.diff(X, n=n)
    return np.concatenate([ np.full(n, np.nan), diffs])

# Calculate first and second differences
first_difference = get_diff(X,1)
second_difference = get_diff(X, n=2)

print("First Difference:", first_difference)
print("Second Difference:", second_difference)
```

```python
# Create a DataFrame
df = pd.DataFrame({'X':X,'first_lag':first_lag,'second_lag':second_lag,'first_difference':first_difference,'second_difference':second_difference})
df
```
